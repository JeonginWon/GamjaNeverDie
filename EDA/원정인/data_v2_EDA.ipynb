```python
import pandas as pd
path_summary = "/content/drive/MyDrive/GamjaNeverDie/EDA/원정인/가맹점_위험지표_요약.csv"
path_available = "/content/drive/MyDrive/GamjaNeverDie/EDA/원정인/available_features_volatility.csv"
path_total = "/content/drive/MyDrive/GamjaNeverDie/EDA/원정인/Total_Data_v1.csv"

summary_df = pd.read_csv(path_summary)
available_df = pd.read_csv(path_available)
total_df = pd.read_csv(path_total)
```

```python
print("✅ summary_df columns:", summary_df.columns.tolist())
print("✅ available_df columns:", available_df.columns.tolist())
print("✅ total_df columns:", total_df.columns.tolist())
```

```python
merged_df = (
    total_df
    .merge(summary_df, on="가맹점구분번호", how="left")
    .merge(available_df, on="가맹점구분번호", how="left")
)

print(f"✅ 병합 완료! shape: {merged_df.shape}")
```
